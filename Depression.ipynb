```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("anthonytherrien/depression-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import glob

# Znajdź pliki CSV
csv_files = glob.glob(os.path.join(path, "*.csv"))

# Wczytaj pierwszy plik CSV
df = pd.read_csv(csv_files[0])

# Wyświetl dane
print(df.head())

# Informacje o zbiorze
print(df.info())

# Liczba wierszy i kolumn
print(df.shape)
```

```python
kolumny_numeryczn = df.select_dtypes(include=['number']).columns
print(kolumny_numeryczn)

```

```python
if len(kolumny_numeryczn) > 0:
  print("Statystyki dla kolumn numerycznych:")
  print(df[kolumny_numeryczn].describe().T.round(2))

```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.boxplot(
    data=df,
    x='Number of Children',
    y='Income'
)

plt.title('Dochód w zależności od liczby dzieci')
plt.xlabel('Liczba dzieci')
plt.ylabel('Dochód')

plt.show()
```

```python
kolumny_nienumeryczne = df.select_dtypes(include=['object']).columns
print(kolumny_nienumeryczne)
if len(kolumny_nienumeryczne) > 0:
  for kolumna in kolumny_nienumeryczne:
    print("f\nKolumna: {kolumna}")

    print(f"Wartości unikalne w kolumnie '{kolumna}':")
    print(df[kolumna].unique())
    print('3 najczestsze wartości')
    print(df[kolumna].value_counts().head(3))
```
